# 逻辑回归API使用
# 逻辑回归处理分类任务：二分类任务，多分类任务。二分类任务使用sigmoid函数，将模型计算的结果转换为成概率；多分类任务使用softmax函数，模型会计算每一个分类的分数，然后softmax函数把分数转换成概率，所有类别概率加起来等于1。

```python
from sklearn.linear_model import LogisticRegression # 逻辑回归模型
```

```python
# solver: 优化算法
#       lbfgs: 拟牛顿法（默认），仅支持L2正则化
#       newton-cg: 牛顿法，仅支持L2正则化
#       liblinear: 坐标下降法，适用于小数据集，支持L1和L2正则化
#       sag: 随机平均梯度下降，适用于大规模数据集，仅支持L2正则化
#       saga: 改进的随机梯度下降，适用于大规模数据集，支持L1、L2和ElasticNet正则化
# penalty: 正则化类型，可选l1、l2和elasticnet
# C: 正则化强度，C越小，正则化强度越大
# class_weight: 类别权重，balanced表示自动平衡类别权重，让模型在训练时更关注少数类，从而减少类别不平衡带来的偏差
model = LogisticRegression(solver='lbfgs', penalty='l2', C=1, class_weight='balanced')
```

# 逻辑回归处理多分类任务

### 注：新版本的sklearn已经不支持在LogisticRegression中设置multi_class参数，所以不支持通过 一对多（OVR）处理多分类任务。LogisticRegression 默认使用 multinomial / softmax 处理多分类任务。如果真的需要使用一对多（OVR）处理多分类任务，可以使用 OneVsRestClassifier 包装 LogisticRegression。


## 1. 一对多（OVR）

```python
from sklearn.multiclass import OneVsRestClassifier # 一对多分类器

model_ovr = OneVsRestClassifier(LogisticRegression())
```
