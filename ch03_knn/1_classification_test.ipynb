# KNN 算法 - 分类任务

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# 1. 准备数据
X = [[2, 1], [3, 1], [1, 4], [2, 6]] # 特征
y = [0, 0, 1, 1] # 标签
```

```python
# 2. 定义模型：KNN分类器实例对象，设置邻居数为2
knn = KNeighborsClassifier(2)
```

```python
# 3. 训练模型
knn.fit(X, y) # fit 训练模型，X 是特征数据，y 是标签数据
```

```python
# 4. 预测新数据的类别
x = [[4, 9]]
x_class = knn.predict(x)
print(x_class)
```

# 画图

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 转成ndarray
X = np.array(X)
y = np.array(y)
print(X.shape)
print(y.shape)
```

```python
# 按类别将点分为两组
X1 = X[y==0]
X2 = X[y==1]
print(X1)
print(X2)
```

```python
colors = ['C0', 'C1']
plt.scatter(X1[:, 0], X1[:, 1], c=colors[0])
plt.scatter(X2[:, 0], X2[:, 1], c=colors[1])

x_color = colors[0] if x_class[0] == 0 else colors[1]
plt.scatter(x[0][0], x[0][1], c=x_color)
plt.show()
```
