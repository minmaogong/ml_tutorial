# 归一化

```python
from sklearn.preprocessing import MinMaxScaler
```

·

```python
# 1. 准备数据（特征）
X = [[2, 1], [3, 1], [1, 4], [2, 6]]
print(X)
```

```python
# 2. 定义归一化缩放器
scaler = MinMaxScaler(feature_range=(0, 1))
```

```python
# 3. 特征转换
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

# 标准化

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
mean = np.mean(X, axis=0) # axis 0: 列  1: 行
std = np.std(X, axis=0)
print( (X - mean) / std)
```
