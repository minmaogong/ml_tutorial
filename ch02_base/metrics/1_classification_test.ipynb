# 分类模型评价指标

## 混淆矩阵

```python
from  sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
import pandas as pd
```

```python
labels = ["猫", "狗"] # 标签 正类：猫 负类：狗
y_true = ["猫", "猫", "猫", "猫", "猫", "猫", "狗", "狗", "狗", "狗"] # 真实值
y_pred = ["猫", "猫", "狗", "猫", "猫", "猫", "猫", "猫", "狗", "狗"] # 预测值
matrix = confusion_matrix(y_true=y_true, y_pred=y_pred, labels=labels) # 计算混淆矩阵 y_true: 真实值 y_pred: 预测值 labels: 标签列表
print(matrix)
```

```python
# 将混淆矩阵转换为DataFrame格式，方便查看
df = pd.DataFrame(matrix, index=labels, columns=labels)
print(df)
```

```python
import  seaborn as sns
sns.heatmap(matrix, annot=True, fmt="d", cmap="Greens") # 绘制混淆矩阵热力图 annot=True: 显示数值 fmt="d": 数值格式为整数 cmap="Greens": 颜色映射为绿色
```

## 准确率（Accuracy）

```python
accuracy = accuracy_score(y_true=y_true, y_pred=y_pred) # 计算准确率  正确预测的比例
print("准确率:", accuracy)
```

## 精确率（Precision）

```python
precision = precision_score(y_true=y_true, y_pred=y_pred, labels=labels, pos_label="猫") # 计算精确率 预测为正例的样本中实际为正例的比例，也叫查准率
print("精确率:", precision)
```

## 召回率（Recall）

```python
recall = recall_score(y_true=y_true, y_pred=y_pred, labels=labels, pos_label="猫")  # 计算召回率 实际为正例的样本中被正确预测为正例的比例，也叫查全率
print("召回率:", recall)
```

## F1分数（F1 Score）

```python
f1 = f1_score(y_true=y_true, y_pred=y_pred, labels=labels, pos_label="猫") # 计算F1分数  精确率和召回率的调和平均数，综合考虑了精确率和召回率
print("F1分数:", f1)
```

## 评估报告

```python
report = classification_report(y_true=y_true, y_pred=y_pred, labels=labels) # 生成分类报告 包含精确率、召回率、F1分数等指标
print(report)
```
