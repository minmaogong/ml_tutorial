```python
# 低方差过滤（Variance Thresholding）是一种简单的特征选择方法，旨在移除那些方差较小的特征。方差较小的特征通常对模型的预测能力贡献较小，因为它们在数据中变化不大，可能是常数或接近常数的特征。
```

```python
import numpy as np
```

```python
# 特征a
a = np.random.randn(100) # randn 返回符合标准正态分布的100个随机数（均值=0，标准差=1）
print(a)
```

```python
print(np.var(a)) # var 计算方差
```

```python
# 生成方差较小的特征
b = np.random.normal(5, 0.1, 100) # normal 生成符合正态分布的随机数，参数分别为均值、标准差和数量
print(b)
```

```python
print(np.var(b)) # 计算方差
```

```python
# 构造输入数据X，有两个特征a和b
X = np.vstack((a, b)).T # vstack 将a和b垂直堆叠，T 转置
print(X)
print(X.shape)
```

## 低方差过滤

```python
from sklearn.feature_selection import VarianceThreshold
```

```python
# 定义方差过滤器，设置阈值为0.01
vt = VarianceThreshold(threshold=0.01) # threshold 设置方差的阈值
```

```python
# 对数据进行过滤（特征选择）
X_filtered = vt.fit_transform(X) # fit_transform 先拟合数据，然后进行变换（把方差小的过滤掉 方差越小，特征与结果越无关）
```

```python
print(X_filtered)
print(X_filtered.shape)
```
