# 相关系数法（Correlation Coefficient）是衡量两个变量之间线性关系强度和方向的统计指标。常用的相关系数有皮尔逊相关系数（Pearson Correlation Coefficient）和斯皮尔曼相关系数（Spearman Correlation Coefficient）。在本章中，我们将重点介绍皮尔逊相关系数。

```python
import pandas as pd
```

```python
# 读取数据
df = pd.read_csv('../../data/advertising.csv')
print(df.shape)
```

```python
# 数据清洗
df.drop(df.columns[0], axis=1, inplace=True) # drop 删除指定的行或列，axis=1 表示删除列，inplace=True 表示直接修改原数据框
df.dropna()
print(df.shape)
```

```python
# 提取特征和标签
X = df.drop('Sales', axis=1) # 提取特征 TV Radio Newspaper
Y = df['Sales'] # 提取标签 Sales
print(X)
print(Y)
```

```python
# 计算相关系数
corr = X.corrwith(Y) # corrwith 计算每个特征与标签之间的相关系数 method='pearson' 默认使用皮尔逊相关系数
print(corr)
```

```python
# 计算相关系数矩阵
corr_matrix = df.corr() # corr 计算数据框中每对特征之间的相关系数矩阵 method='pearson' 默认使用皮尔逊相关系数
print(corr_matrix)
```

```python
# 可视化
import  seaborn as sns
```

```python
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm') # heatmap 绘制热力图，annot=True 显示数值，fmt='.2f' 设置数值格式为两位小数，cmap='coolwarm' 设置颜色映射
```
