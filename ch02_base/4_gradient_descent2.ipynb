# 梯度下降法 案例2

```python
# 定义目标函数（模拟损失函数）
def J(x):
    return (x ** 2 - 2) ** 2

# 定义梯度函数（模拟损失函数的导数）
def grad_f(x):
    return 4 * x ** 3 - 8 * x
```

```python
# 记录参数更新及损失值轨迹
x_list = []
y_list = []
```

```python
# 1. 初始化参数
x = 1
# 设置学习率（超参数）
learning_rate = 0.1
```

```python
# 4. 迭代更新参数
for i in range(100):
    # 记录当前点坐标
    y = J(x)
    x_list.append(x)
    y_list.append(y)
    print(f"x: {x}, J(x): {y}")

    # 2. 计算梯度
    grad = grad_f(x)
    # 3. 更新参数
    x = x - learning_rate * grad
```

```python
print(x)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(0.9, 1.6, 0.01)
```

```python
plt.plot(x, J(x), label='f(x)')
plt.plot(x_list, y_list, c='r')
plt.scatter(x_list, y_list, color='red')
plt.show()
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,4))
ax[0].plot(x, J(x))
ax[0].plot(x_list, y_list, c='r')
ax[0].scatter(x_list, y_list, color='red')

x_list1 = x_list[1:]
y_list1 = y_list[1:]

x = np.arange(1.399, 1.425, 0.001) # 细化x轴范围
ax[1].plot(x, J(x))
ax[1].plot(x_list1, y_list1, c='r')
ax[1].scatter(x_list1, y_list1, color='red')

plt.show()
```
