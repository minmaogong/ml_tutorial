# 梯度下降法 案例1

```python
# 定义目标函数（模拟损失函数）
def f(x):
    return x ** 2

# 定义梯度函数（模拟损失函数的导数）
def grad_f(x):
    return 2 * x
```

```python
# 记录参数更新及损失值轨迹
x_list = []
y_list = []
```

```python
# 1. 初始化参数
x = 1
# 设置学习率（超参数）
learning_rate = 0.1
```

```python
# 4. 迭代更新参数
for i in range(100):
    # 记录当前点坐标
    y = f(x)
    x_list.append(x)
    y_list.append(y)
    print(f"x: {x}, f(x): {y}")

    # 2. 计算梯度
    grad = grad_f(x)
    # 3. 更新参数
    x = x - learning_rate * grad
```

```python
print(x)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(-1.2, 1.2, 0.01)
```

```python
plt.plot(x, f(x), label='f(x)')
plt.plot(x_list, y_list, c='r')
plt.scatter(x_list, y_list, color='red')
plt.show()
```
